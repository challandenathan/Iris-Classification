DATA EXPLORATION 

```python
import pandas as pd
train_df = pd.read_csv('data/train(2).csv')
```

PLOT OF THE FEATURES

```python
import matplotlib.pyplot as plt
X_train = train_df.drop(columns = ["id","target"])
Y_train = train_df['target']
print(X_train.head())
plt.scatter(X_train['sepal length (cm)'],Y_train)
plt.scatter(X_train['sepal width (cm)'],Y_train)
plt.scatter(X_train['petal length (cm)'],Y_train)
plt.scatter(X_train['petal width (cm)'],Y_train)
plt.show
```
